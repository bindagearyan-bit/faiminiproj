## 📖 How This Google Colab Backend Works:
1. **Cloud Execution:** This notebook loads the trained Scikit-Learn classification pipeline on Google's cloud servers.
2. **Simultaneous Connection:** When users move sliders or type values on the website, the website sends an AJAX request directly to this Colab backend.
3. **Live Output in Colab:** This notebook prints each incoming patient's vitals, calculates the risk percentage, generates the dark-mode Seaborn heatmap / Matplotlib plot, and streams it back to the website in real time (~20ms latency)!
